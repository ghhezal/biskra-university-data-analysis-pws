```python
plt.figure(figsize = (8, 5))
sns.boxplot(x = df["Grade"])
plt.xlabel("Grade")
plt.title("Boxplot of Student Grades")
plt.show()
```

```python
Q1 = df["Grade"].quantile(0.25)
Q3 = df["Grade"].quantile(0.75)
IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = df[
    (df["Grade"] < lower_bound) |
    (df["Grade"] > upper_bound)
]
```

```python
correlation_matrix = df[[
    "Age",
    "Study_Hours",
    "Grade",
    "Absences"
]].corr()
```

```python
correlation_matrix
```

```python
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

**strong positive correlation:**

- Study_Hours and Grade.

- Absences and Age.

**strong negative correlation**:

- Study_Hours and Absences.

**weak correlation**

- Age and Grade.

- Age and Study_Hours.
